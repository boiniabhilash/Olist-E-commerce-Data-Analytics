```python
def create_prompt(question):
    prompt = f"""
You are an AI Business Analyst.

Business Question:
{question}

Olist Business Evidence:
{knowledge_base_dynamic}

Answer using:
1. Key finding
2. Evidence
3. Possible factors
4. Business recommendation

Rules:
- Use only the evidence provided.
- Do not claim correlation means causation.
- Clearly separate facts from assumptions.
- Do not invent numbers.
"""
    return prompt
```

```python
def ai_business_answer(question):
    question = question.lower()

    if "low reviews" in question:
        answer = """
Key finding:
Delivery performance appears to be associated with customer satisfaction.

Evidence:
- Average delivery time was 12.09 days.
- There were 6,534 late orders.
- Delivery days and review score had a correlation of -0.334.
- 12,350 orders received a review score of 1 or 2.

Possible factors:
Longer delivery times may be associated with lower review scores.

Business recommendation:
Investigate delivery delays and identify opportunities to improve
delivery reliability.

Note:
Correlation does not prove causation.
"""

    elif "delivery" in question:
        answer = """
Key finding:
Delivery performance is an important area for investigation.

Evidence:
- Average delivery time was 12.09 days.
- There were 6,534 late orders.
- Average days late was 10.62 days.

Possible factors:
Late deliveries may be associated with customer dissatisfaction.

Business recommendation:
Investigate the causes and locations of delivery delays.

Note:
The analysis does not prove causation.
"""

    elif "improv" in question or "focus" in question:
        answer = """
Key finding:
Delivery performance and customer satisfaction should be investigated.

Evidence:
- Average delivery time was 12.09 days.
- There were 6,534 late orders.
- Delivery days and review score correlation was -0.334.
- 12,350 orders received low review scores.

Business recommendation:
1. Investigate causes of late deliveries.
2. Identify areas with higher delivery delays.
3. Monitor low-review orders.
4. Test whether delivery improvements affect satisfaction.

Note:
These findings do not prove causation.
"""

    elif "predict" in question or "model" in question:
        answer = """
Key finding:
The balanced Logistic Regression model identified some low-review orders,
but its performance was limited.

Evidence:
- Baseline model recall: 11.98%.
- Balanced model recall: 53.77%.
- Balanced model F1 score: 33.50%.

Interpretation:
The balanced model detected more actual low-review orders than the
baseline model, but it also produced false positive predictions.

Business recommendation:
Use the model as an early-warning tool for further investigation,
not as a final automated decision-making system.

Note:
The model is a baseline analysis and is not production-ready.
"""

    else:
        answer = """
No specific analysis has been created for this question yet.
"""

    return answer
```

## AI Response Evaluation Checklist

Every AI-generated business answer should be checked for:

- **Evidence:** Does the answer use the Olist analysis?
- **Accuracy:** Are the numbers consistent with the analysis?
- **Relevance:** Does it directly answer the business question?
- **Business value:** Does it provide a useful business recommendation?
- **Uncertainty:** Does it distinguish facts from assumptions?
- **Causation:** Does it avoid treating correlation as causation?
- **Hallucination:** Does it avoid inventing information?
- **Model limitations:** Does it acknowledge predictive-model limitations when relevant?

```python
def check_ai_answer(answer):
    checks = {
        "Has answer": len(answer) > 0,
        "Has evidence": "evidence" in answer.lower(),
        "Has recommendation": "recommendation" in answer.lower(),
        "Mentions limitations": "limitation" in answer.lower()
    }

    for check, result in checks.items():
        print(check + ":", result)
```
